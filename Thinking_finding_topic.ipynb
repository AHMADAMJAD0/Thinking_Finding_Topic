# **Proposed Research Thesis: Investigating the Impact of LoRA-Family Fine-Tuning Methods on the Compressibility and Mobile Deployment Performance of Large Language Models**

**I will compare different fine-tuning methods applied to the same language model, then investigate how those fine-tuned models perform on real mobile phones.**

Main research question:
How do different LoRA-family fine-tuning methods influence the compressibility, accuracy, and inference efficiency of large language models deployed on memory-constrained smartphones?

Sub-questions:
- RQ1: Which fine-tuning method retains the highest accuracy after INT4, 3-bit, and 2-bit quantization?
- RQ2: Which fine-tuned and compressed models can successfully run on smartphones with 8 GB and 12 GB RAM?
- RQ3: How do the methods compare in inference speed, peak RAM usage, and energy consumption after deployment?
- RQ4: Which combination of fine-tuning method and quantization level offers the best accuracy–memory–latency trade-off?

You will find which combination gives the best results, such as:
- Best accuracy after compression
- Lowest inference memory
- Fastest response generation
- Best performance on 8 GB versus 12 GB phones
- Greatest compression without unacceptable accuracy loss
Importantly, you may discover that no single fine-tuning method is best for every metric.
